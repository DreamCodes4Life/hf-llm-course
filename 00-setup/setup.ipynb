## Project folder (course: *Using a Python virtual environment*)

The course creates a project folder in your home directory (`~` is your home folder):

```bash

```

Here the folder lives in the root of this repo instead, so the commands drop the `~/`.

Each `!` line runs in its own short-lived shell, so a `!cd` would be forgotten by the next line. `%%bash` runs the whole cell as one bash script, like a terminal. It doesn't change the notebook's own working folder (that would be `%cd`).

````python
%%bash
python --version                        # which Python is the "main" one here
cd "$(git rev-parse --show-toplevel)"   # start at the root of the repo
mkdir -p transformers-course            # -p: no error if it already exists
cd transformers-course
pwd
````

## First pipeline (from Chapter 1)

Downloads a small sentiment model the first time (cached in `~/.cache/huggingface`).

````python
from transformers import pipeline

classifier = pipeline('sentiment-analysis', device=0 if torch.cuda.is_available() else -1)
classifier([
    "I've been waiting for a HuggingFace course my whole life.",
    'I hate this so much!',
])
````
